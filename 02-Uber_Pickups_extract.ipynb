# Extract data

```python
import zipfile
print('Extract')
try:
    with zipfile.ZipFile('uber-trip-data.zip', 'r') as z:
        print("Contenu de l'archive :")
        for nom_fichier in z.namelist():
            print(f"  - {nom_fichier}")
except FileNotFoundError:
    print("L'archive 'uber-trip-data.zip' n'existe pas encore. Nous allons en créer une plus tard.")
```

L'archive contient plusieurs fichiers csv et une seconde archive.

```python
import os
# Define the path to the zip file and the extraction directory

with zipfile.ZipFile('uber-trip-data.zip', mode = 'r') as z:
    z.extractall(path='data')
print('uber-trip-data.zip extracted...')
```

```python
import shutil
for f in os.listdir("data/uber-trip-data"):
    shutil.move(os.path.join("data/uber-trip-data", f), "data")
```

```python

unused_folds = ['data/__MACOSX', 'data/uber-trip-data']
for f in unused_folds:
    shutil.rmtree(f)
```
